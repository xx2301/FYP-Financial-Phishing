```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime
import pandas as pd

tasks = [
    ("Literature Review", "2026-09-26", "2026-10-15"),
    ("Dataset Preparation & Labelling", "2026-10-01", "2026-10-20"),
    ("Feature Engineering & Splitting", "2026-10-19", "2026-11-02"),
    ("Baseline Model Training", "2026-11-02", "2026-11-20"),
    ("SHAP & Robustness Testing", "2026-11-15", "2026-11-20"),
    ("FEMRIS Framework Development", "2026-11-15", "2026-11-20"),
    ("Thesis Writing & Revision", "2026-10-05", "2026-11-26"),
]

data = []
for t in tasks:
    start = datetime.strptime(t[1], "%Y-%m-%d")
    end = datetime.strptime(t[2], "%Y-%m-%d")
    data.append((t[0], start, end))

fig, ax = plt.subplots(figsize=(10, 5))
colors = ['#4CAF50', '#2196F3', '#FF9800', '#F44336', '#9C27B0', '#00BCD4', '#795548']

for i, (name, start, end) in enumerate(data):
    ax.barh(name, (end - start).days, left=start, color=colors[i % len(colors)], edgecolor='black', height=0.5)
    
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax.xaxis.set_major_locator(mdates.MonthLocator())
plt.xticks(rotation=45)

ax.invert_yaxis()
ax.set_xlabel("Timeline", fontsize=12)
ax.set_title("Project Management Gantt Chart", fontsize=14, fontweight='bold')
ax.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()

#plt.savefig("gantt_chart.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
