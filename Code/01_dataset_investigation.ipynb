```python
CORE_TERMS = ['bank', 'banking', 'account', 'payment', 'payments', 'invoice', 'billing', 'refund', 'card', 'debit', 'credit', 'credit card', 'mastercard', 'visa', 'transaction', 'transfer', 'wire', 'wallet', 'payroll', 'salary', 'statement', 'balance', 'tax', 'irs', 'ato', 'kyc', 'aml', 'swift', 'otp', 'security code', 'verification code', 'one time password', 'direct debit', 'bill', 'receipt', 'purchase', 'merchant', 'beneficiary', 'remittance', 'overdue', 'loan', 'mortgage', 'insurance', 'superannuation']
BRANDS = ['paypal', 'stripe', 'wise', 'western union', 'payoneer', 'venmo', 'cash app', 'bank of america', 'chase', 'wells fargo', 'citibank', 'hsbc', 'barclays', 'lloyds', 'natwest', 'commonwealth bank', 'commbank', 'nab', 'anz', 'westpac', 'amex', 'american express', 'mastercard', 'visa']
CONTEXT_PHRASES = ['verify your account', 'account suspended', 'account locked', 'unusual activity', 'suspicious login', 'confirm your identity', 'payment failed', 'invoice attached', 'update your billing', 'secure message', 'transaction failed', 'unauthorised transaction', 'reset your password', 'confirm your payment', 'pending refund', 'your statement is ready', 'click below to verify', 'login to continue', 'validate your account', 'payment confirmation', 'direct deposit', 'wire transfer']
ACTIONS = ['verify', 'confirm', 'update', 'login', 'sign in', 'reset', 'unlock', 'suspend', 'limited', 'urgent', 'immediately', 'action required', 'click here', 'open attachment', 'respond now', 'security alert']
```

```python
def calculate_finance_score(text):
    if not isinstance(text, str):
        return 0, False
    text_lower = text.lower()
    score = 0
    has_finance_signal = False
    
    if any(kw in text_lower for kw in CORE_TERMS):
        score += 1
        has_finance_signal = True
    if any(kw in text_lower for kw in BRANDS):
        score += 2
        has_finance_signal = True
    if any(kw in text_lower for kw in CONTEXT_PHRASES):
        score += 3
        has_finance_signal = True
    if any(kw in text_lower for kw in ACTIONS):
        score += 1
        
    return score, has_finance_signal

print("Scanning all URLs according to the paper's method...")
scan_results = df['url'].apply(calculate_finance_score)

df['finance_score'] = scan_results.apply(lambda x: x[0])
df['has_finance_signal'] = scan_results.apply(lambda x: x[1])

df['is_financial'] = (
    (df['finance_score'] >= 2) & 
    (df['has_finance_signal'] == True)
)

print("\n=== Final financial tag distribution (Pure Paper Method) ===")
print(df.groupby(['is_financial', 'label']).size())

print("\n=== Top 20 targets in finance (Pure Paper Method) ===")
print(df[df['is_financial'] == True]['target'].value_counts().head(20))

df_to_save = df.drop(columns=['html'], errors='ignore')
df_to_save.to_csv("phreshphish_35k_pure_paper_baseline.csv", index=False)
print("\n✅ Saved to phreshphish_35k_pure_paper_baseline.csv")
```

```python
def draw(d, n=50, seed=42):
    s = pd.concat([g.sample(min(len(g), n // 2), random_state=seed)
                   for _, g in d.groupby('label')])
    if len(s) < n:
        rest = d.drop(s.index)
        s = pd.concat([s, rest.sample(min(len(rest), n - len(s)), random_state=seed)])
    return s

parts = []
for d in (fin, non):
    for src, g in d.groupby('auto_source'):
        parts.append(draw(g).assign(stratum=f"{src}_{g.auto_label.iloc[0]}"))
sample = pd.concat(parts)

def fetch_html(path, ids):
    out = [c[c.sha256.isin(ids)] for c in
           pd.read_csv(path, usecols=['sha256', 'html'], chunksize=1000)]
    return pd.concat(out)

html = pd.concat([fetch_html('phreshphish_financial.csv', set(sample.sha256)),
                  fetch_html('phreshphish_non_financial.csv', set(sample.sha256))])
sample = sample.merge(html, on='sha256')
```

```python
import pandas as pd
from statsmodels.stats.proportion import proportion_confint

done = pd.read_excel("label_validation_blind.xlsx")
full = pd.read_csv("label_validation_full.csv")

done['manual_label'] = done['manual_label'].astype(str).str.strip().str.lower()

m = full.merge(done[['id', 'manual_label']], on='id')
m['agree'] = m.manual_label == m.auto_label

rows = []
for s, g in m.groupby('stratum'):
    k, n = int(g.agree.sum()), len(g)
    lo, hi = proportion_confint(k, n, method='wilson')
    rows.append((s, n, k, f"{k/n:.1%}", f"[{lo:.1%}, {hi:.1%}]"))
print(pd.DataFrame(rows, columns=['stratum', 'n', 'agree', 'rate', '95% CI']))
print("overall:", int(m.agree.sum()), "/", len(m))

print(m.groupby(['stratum', 'label']).agree.agg(['sum', 'count', 'mean']))

print(pd.crosstab(m.auto_label, m.manual_label))

m[~m.agree][['id', 'stratum', 'label', 'auto_label', 'manual_label', 'url', 'target']] \
    .to_csv("label_disagreements.csv", index=False)
print("disagreements:", (~m.agree).sum())
```

```python
import re, pandas as pd

done = pd.read_excel("label_validation_blind.xlsx")
done['manual_label'] = done['manual_label'].astype(str).str.strip().str.lower()
full = pd.read_csv("label_validation_full.csv")
m = full.merge(done[['id', 'manual_label']], on='id')
d = m[m.manual_label != m.auto_label].copy()
print(len(d)) # might be 94

def cause(r):
    u = str(r.url).lower()
    if r.manual_label == 'uncertain':
        return "Insufficient page content"
    if r.stratum == 'lexicon_financial':
        core = [k for k in CORE_TERMS if k in u]
        brand = [k for k in BRANDS if k in u]
        word = lambda k: re.search(r'(?<![a-z])' + re.escape(k) + r'(?![a-z])', u)
        if not any(word(k) for k in core + brand):
            return "Substring match in unrelated word"
        if core == ['account'] and not brand:
            return "Generic term ``account''"
        return "Financial term, page not a financial service"
    if r.stratum == 'lexicon_non-financial':
        return "No financial term in URL; page content financial"
    if r.stratum == 'uncertain_uncertain':
        return "Discarded as uncertain; page content "+r.manual_label
    return "Target ambiguous or page not matching target"

d['cause'] = d.apply(cause, axis=1)
print(d.groupby(['stratum', 'cause']).size())

abbr = {'lexicon_financial': 'Lex-fin', 'lexicon_non-financial': 'Lex-nonfin',
        'whitelist_financial': 'Whitelist', 'blacklist_non-financial': 'Blacklist',
        'uncertain_uncertain': 'Uncertain'}
rows = [f"    {r.id} & {r.auto_label} & {r.manual_label} & {abbr[r.stratum]} & "
        f"{r.cause}; no change \\\\" for r in d.sort_values(['stratum', 'id']).itertuples()]
open("disagreement_rows.tex", "w", encoding="utf-8").write("\n".join(rows))
```
